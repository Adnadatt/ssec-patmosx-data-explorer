```python
import pandas as pd

l3 = pd.read_pickle('l3_files.pickle')
print("L3 max date:", l3['date'].max())
print("L3 min date:", l3['date'].min())

# per product, since coverage may differ
print(l3.groupby('key')['date'].agg(['min', 'max']))

# per platform too, if you want to see which satellites have the most recent data
print(l3.groupby('platform')['date'].agg(['min', 'max']).sort_values('max', ascending=False))
```
